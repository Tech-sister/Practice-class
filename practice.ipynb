```python
#import libraries
import numpy as np
```

```python
#working with numpy array
arr = np.array([40,24,37,45,78,90,50])
print("1D array:", arr)
print("\n Average number:", arr.mean())
```

```python
#Indexing
print(arr[2])
print(arr[-4])
print(arr[5])
```

```python
#slicing
print(arr[1:5])
print(arr[:4])
print(arr[9:])
print(arr[-2:-1])
print(arr[8:2])
print(arr[-1:-6])
print(arr[:9])
```

```python
#2D array with indexing and slicing
arr2 = np.array([
    [2,3,4],
    [5,6,7],
    [1,8,9]
])

print(arr2)
print(arr2[0,1])
print(arr2[1, :])
print(arr2[1:2, 2])
print(arr2[2, :2])
print(arr2[1:2, 0:2])
                
```

```python
#1D to 2D array
arr3 = np.arange(1,65)
reshape = arr3.reshape(8, 8)
print("Array numbers:",arr3)
print("Reshape 1D array into 2D array:\n", reshape)

flatten = reshape.flatten()
print("Flatten array:", flatten)
```

```python
#working with mathematics operation
arr4 = np.array([11,22,33,44,55])
arr5 = np.array([10,20,30,40,50])

#basic math operation
print(arr4 + arr5)
print(arr4 - arr5)
print(arr4 * arr5)
print(arr4 / arr5)

#scaler operation
print(arr5 ** 4)
print(arr5 // 7)
print(arr5 % 9)
```

```python
#working with dot product
matrix_1 = np.array([
    [5, 2],
    [3, 6]
])

matrix_2 = np.array([
    [9, 4],
    [7, 1]
])

dot_product = np.dot(matrix_1, matrix_2)

multiply_arr = matrix_1 * matrix_2
divide_arr = matrix_1 / matrix_2
print("Dot Product output:\n", dot_product)
print("Multiply arr:", multiply_arr)
print("Divide arr:", divide_arr)
```
